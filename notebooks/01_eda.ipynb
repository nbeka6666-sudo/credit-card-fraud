### Feature Engineering

```python
df['hour'] = (df['Time'] // 3600) % 24
df['log_amount'] = np.log1p(df['Amount'])
df = df.drop(columns=['Time', 'Amount'])

print(df.shape)
df.head()
```

## Train / Validation / Test Split

**Critical:** Split is done **before** SMOTE and scaling to avoid data leakage.
`stratify=y` keeps the fraud ratio consistent across all sets.

```python
from sklearn.model_selection import train_test_split

df_full_train, df_test = train_test_split(
    df, test_size=0.2, random_state=1, stratify=df['Class']
)
df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=1, stratify=df_full_train['Class']
)

df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)

y_train = df_train['Class'].values
y_val = df_val['Class'].values
y_test = df_test['Class'].values

del df_train['Class']
del df_val['Class']
del df_test['Class']

print("Train:", df_train.shape, y_train.shape, f"fraud: {y_train.sum()}")
print("Val:  ", df_val.shape, y_val.shape, f"fraud: {y_val.sum()}")
print("Test: ", df_test.shape, y_test.shape, f"fraud: {y_test.sum()}")
```

## Feature Scaling

V1–V28 are PCA components (already standardized). Only `log_amount` and `hour` need scaling.
**Critical:** `fit` on train only — no leakage.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
cols_to_scale = ['log_amount', 'hour']

df_train[cols_to_scale] = scaler.fit_transform(df_train[cols_to_scale])
df_val[cols_to_scale] = scaler.transform(df_val[cols_to_scale])
df_test[cols_to_scale] = scaler.transform(df_test[cols_to_scale])

print(df_train[cols_to_scale].describe().round(2))
```

## Handling Class Imbalance with SMOTE

**Critical:** SMOTE applied **only on train** — never on val/test.

```python
from imblearn.over_sampling import SMOTE

print("Before SMOTE:")
print(f"  Normal: {(y_train == 0).sum()}")
print(f"  Fraud:  {(y_train == 1).sum()}")

smote = SMOTE(random_state=1)
X_train_res, y_train_res = smote.fit_resample(df_train, y_train)

print("\nAfter SMOTE:")
print(f"  Normal: {(y_train_res == 0).sum()}")
print(f"  Fraud:  {(y_train_res == 1).sum()}")
print(f"\nX_train_res shape: {X_train_res.shape}")
```

## Baseline Model: Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    precision_score, recall_score, f1_score,
    roc_auc_score, average_precision_score,
    confusion_matrix
)

lr = LogisticRegression(max_iter=1000, random_state=1)
lr.fit(X_train_res, y_train_res)

y_pred_proba_lr = lr.predict_proba(df_val)[:, 1]
y_pred_lr = (y_pred_proba_lr > 0.5).astype(int)

print("Logistic Regression — Validation Metrics")
print("=" * 45)
print(f"Precision: {precision_score(y_val, y_pred_lr):.4f}")
print(f"Recall:    {recall_score(y_val, y_pred_lr):.4f}")
print(f"F1:        {f1_score(y_val, y_pred_lr):.4f}")
print(f"ROC-AUC:   {roc_auc_score(y_val, y_pred_proba_lr):.4f}")
print(f"PR-AUC:    {average_precision_score(y_val, y_pred_proba_lr):.4f}")
print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_pred_lr))
```

### Threshold Tuning (F1-optimal)

```python
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(y_val, y_pred_proba_lr)

f1_scores = 2 * precision * recall / (precision + recall + 1e-10)
best_idx = np.argmax(f1_scores[:-1])
best_threshold_lr = thresholds[best_idx]

y_pred_lr_best = (y_pred_proba_lr > best_threshold_lr).astype(int)

print(f"Best threshold: {best_threshold_lr:.4f}")
print(f"F1:             {f1_score(y_val, y_pred_lr_best):.4f}")
print(f"Precision:      {precision_score(y_val, y_pred_lr_best):.4f}")
print(f"Recall:         {recall_score(y_val, y_pred_lr_best):.4f}")
```

```python
# 1. Exact threshold value without rounding
print("Exact threshold:", repr(best_threshold_lr))
print(f"Threshold with 8 decimals: {best_threshold_lr:.8f}")

# 2. Distribution of predicted probabilities
import matplotlib.pyplot as plt

plt.hist(y_pred_proba_lr, bins=50)
plt.title("Predicted Probabilities (LogReg + SMOTE)")
plt.xlabel("Probability")
plt.ylabel("Count")
plt.yscale('log')
plt.show()
```

### Threshold near 1.0 — Explained

The optimal F1 threshold for Logistic Regression is **~0.9999** — practically 1.0.

**Why:** LR after SMOTE produces U-shaped probabilities — most predictions are
either near 0 or near 1, with few values in between. The histogram confirms this.

**What it means:**
- The model is highly confident in its predictions (linear + balanced data)
- Threshold 0.9999 is valid — not a bug
- Metrics (F1, Precision, Recall) are correct

**Implication:** Tree-based models (RF, XGBoost) will produce smoother probabilities
and a more "normal" threshold. LR is kept as a baseline only.

## Random Forest

```python
precision, recall, thresholds = precision_recall_curve(y_val, y_pred_proba_rf)

f1_scores = 2 * precision * recall / (precision + recall + 1e-10)
best_idx = np.argmax(f1_scores[:-1])
best_threshold_rf = thresholds[best_idx]

y_pred_rf_best = (y_pred_proba_rf > best_threshold_rf).astype(int)

print(f"Best threshold: {best_threshold_rf:.4f}")
print(f"F1:             {f1_score(y_val, y_pred_rf_best):.4f}")
print(f"Precision:      {precision_score(y_val, y_pred_rf_best):.4f}")
print(f"Recall:         {recall_score(y_val, y_pred_rf_best):.4f}")
```

## XGBoost

```python
from xgboost import XGBClassifier

xgb = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    random_state=1,
    n_jobs=-1,
    eval_metric='logloss'
)
xgb.fit(X_train_res, y_train_res)

y_pred_proba_xgb = xgb.predict_proba(df_val)[:, 1]
y_pred_xgb = (y_pred_proba_xgb > 0.5).astype(int)

print("XGBoost (SMOTE) — Validation Metrics")
print("=" * 45)
print(f"Precision: {precision_score(y_val, y_pred_xgb):.4f}")
print(f"Recall:    {recall_score(y_val, y_pred_xgb):.4f}")
print(f"F1:        {f1_score(y_val, y_pred_xgb):.4f}")
print(f"ROC-AUC:   {roc_auc_score(y_val, y_pred_proba_xgb):.4f}")
print(f"PR-AUC:    {average_precision_score(y_val, y_pred_proba_xgb):.4f}")
print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_pred_xgb))
```

```python
precision, recall, thresholds = precision_recall_curve(y_val, y_pred_proba_xgb)

# F1 formula: 2 * P * R / (P + R). +1e-10 avoids division by zero
f1_scores = 2 * precision * recall / (precision + recall + 1e-10)

# Find threshold that maximizes F1 (skip last element — no matching threshold)
best_idx = np.argmax(f1_scores[:-1])
best_threshold_xgb = thresholds[best_idx]

y_pred_xgb_best = (y_pred_proba_xgb > best_threshold_xgb).astype(int)

print("XGBoost — Validation Metrics (F1-optimal threshold)")
print("=" * 55)
print(f"Best threshold: {best_threshold_xgb:.4f}")
print(f"F1:             {f1_score(y_val, y_pred_xgb_best):.4f}")
print(f"Precision:      {precision_score(y_val, y_pred_xgb_best):.4f}")
print(f"Recall:         {recall_score(y_val, y_pred_xgb_best):.4f}")
```

### XGBoost: SMOTE vs scale_pos_weight

Compare two approaches for handling class imbalance:
- **SMOTE**: synthetic fraud examples in train
- **scale_pos_weight**: native XGBoost parameter, no synthetic data

```python
# Model 1: XGBoost trained on SMOTE-resampled data
xgb_smote = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    random_state=1,
    n_jobs=-1,
    eval_metric='logloss'
)
xgb_smote.fit(X_train_res, y_train_res)
y_pred_proba_smote = xgb_smote.predict_proba(df_val)[:, 1]

# Model 2: XGBoost with scale_pos_weight, trained on original data
scale = (y_train == 0).sum() / (y_train == 1).sum()

xgb_scale = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    scale_pos_weight=scale,
    random_state=1,
    n_jobs=-1,
    eval_metric='logloss'
)
xgb_scale.fit(df_train, y_train)
y_pred_proba_scale = xgb_scale.predict_proba(df_val)[:, 1]

print(f"scale_pos_weight = {scale:.2f}")
```

```python
def best_f1_threshold(y_true, y_proba):
    precision, recall, thresholds = precision_recall_curve(y_true, y_proba)
    f1_scores = 2 * precision * recall / (precision + recall + 1e-10)
    best_idx = np.argmax(f1_scores[:-1])
    return thresholds[best_idx], f1_scores[best_idx]

# SMOTE
t_smote, _ = best_f1_threshold(y_val, y_pred_proba_smote)
y_pred_smote_best = (y_pred_proba_smote > t_smote).astype(int)

# scale_pos_weight
t_scale, _ = best_f1_threshold(y_val, y_pred_proba_scale)
y_pred_scale_best = (y_pred_proba_scale > t_scale).astype(int)

print("Comparison on validation set")
print("=" * 60)
print(f"{'Method':<25} {'Threshold':>10} {'Precision':>10} {'Recall':>10} {'F1':>10}")
print("-" * 60)

for name, threshold, y_pred_b in [
    ("SMOTE", t_smote, y_pred_smote_best),
    ("scale_pos_weight", t_scale, y_pred_scale_best),
]:
    p = precision_score(y_val, y_pred_b)
    r = recall_score(y_val, y_pred_b)
    f1 = f1_score(y_val, y_pred_b)
    print(f"{name:<25} {threshold:>10.4f} {p:>10.4f} {r:>10.4f} {f1:>10.4f}")
```

**Decision:** Use **scale_pos_weight** — it outperforms SMOTE on all metrics
(F1, Precision, Recall). `scale_pos_weight` avoids synthetic examples and
gives a cleaner model. Final metrics are shown in the Comparison and Test Set
sections below.
